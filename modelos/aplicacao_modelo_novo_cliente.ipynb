# Funções

```python
# função p criação e clusterização de novo cliente

def clusterizar_novo_cliente(cluster, df_original, modelo_kproto, restrito=True):

    # opção para usar apenas um cluster como base
    if restrito:
        # Define cluster aleatório entre 0 e 5
        df_original = df_original[df_original['CLUSTER_5'] == cluster]

    # cliente sample
    df_original = df_original.sample(n=1, random_state=42)

    # removendo colunas de clusterização
    df_original.drop(columns=['CLUSTER_3', 'CLUSTER_4', 'CLUSTER_5', 'CLUSTER_6'], inplace=True)

    print(df_original)

    # tratamento valores
    df3, df4, df_novo_cliente_1 = trata_df_final(df_original, normalizar = False)

    print(df_novo_cliente_1)

    # Separando colunas categoricas
    categorical_idx = trata_colunas_categoricas(df_novo_cliente_1)

    # definindo cluster
    cluster_previsto = modelo_kproto.predict(
        X=df_novo_cliente_1,
        categorical=categorical_idx
    )

    print(cluster_previsto)

    cluster_resultado = cluster_previsto[0]

    return cluster_resultado, df_novo_cliente_1
```

## PCA

```python
# 1. Supondo que seu DataFrame se chama 'df_novo_cliente_1_clusterizado'
# 2. Defina qual é a coluna de cluster

# 3. Separe as features (X) e os clusters (y)
# Exclua a coluna de clusterização (e outras colunas irrelevantes como IDs) do conjunto X
X = df_unido.drop(columns=['ID_CLIENTE'], errors='ignore')
# Aplicando One-Hot Encoding em todas as colunas categóricas
X_encoded = pd.get_dummies(X)

from sklearn.preprocessing import StandardScaler

# 3.1 Escalonamento/Normalização das features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_encoded)
```

### 2D

```python
from sklearn.decomposition import PCA

# 2.2 Configuração e aplicação do t-SNE (AGORA COM 2 COMPONENTES)
# - n_components=2: para visualização 2D
pca2 = PCA(n_components=2)
X_pca2 = pca2.fit_transform(X_scaled)

# 3.3 Atribuição direta (Se a ordem for garantida)
df_unido['pca_C1_2D'] = X_pca2[:, 0]  # Primeira coluna de X_pca2
df_unido['pca_C2_2D'] = X_pca2[:, 1]  # Segunda coluna de X_pca2
```

### 3D

```python
# 2.2 Configuração e aplicação do t-SNE (AGORA COM 2 COMPONENTES)
# - n_components=2: para visualização 2D
pca3 = PCA(n_components=3)
X_pca3 = pca3.fit_transform(X_scaled)

# 3.3 Atribuição direta (Se a ordem for garantida)
df_unido['pca_C1_3D'] = X_pca3[:, 0]  # Primeira coluna de X_pca2
df_unido['pca_C2_3D'] = X_pca3[:, 1]  # Segunda coluna de X_pca2
df_unido['pca_C3_3D'] = X_pca3[:, 2]  # Segunda coluna de X_pca2
```

## Visualização 2D

```python
import plotly.express as px

COLUNA_CLUSTER = 'CLUSTER_5'

# A coluna de cluster (COLUNA_CLUSTER) precisa ser do tipo string/object para o Plotly
# Se a sua coluna de cluster (ex: 'CLUSTER_5') for numérica (int/uint), converta-a:
df_unido[COLUNA_CLUSTER] = df_unido[COLUNA_CLUSTER].astype(str)

# --- Criação do Gráfico Interativo ---
fig = px.scatter(
    df_unido,
    x="pca_C1_2D",
    y="pca_C2_2D",
    # Define a coluna que será usada para colorir os pontos (os clusters)
    color='Cliente_Novo',
    # Você pode usar a paleta "hls" do Plotly, ou outras
    color_discrete_sequence=px.colors.qualitative.Alphabet,
    # Texto que aparece ao passar o mouse sobre os pontos
    hover_data=['Cliente_Novo'],
    title=f'Visualização Interativa de Clusters com t-SNE do Cluster {cluster_resultado}'
)

# --- Personalização e Layout ---
fig.update_layout(
    xaxis_title='Componente t-SNE 1',
    yaxis_title='Componente t-SNE 2',
    height=600,  # Altura da figura
    width=800    # Largura da figura
)

# Adiciona uma grade simples (opcional)
fig.update_xaxes(showgrid=True, gridwidth=1, gridcolor='LightGrey')
fig.update_yaxes(showgrid=True, gridwidth=1, gridcolor='LightGrey')

# Mostra o gráfico (será interativo no Jupyter Notebook/Colab)
fig.show()
```

## Visualização 3D

```python
# Garante que o Plotly trate a coluna de cluster como categoria (string)
df_unido['Cliente_Novo'] = df_unido['Cliente_Novo'].astype(str)

# --- Criação do Gráfico 3D Interativo ---
fig = px.scatter_3d(
    df_unido,
    x="pca_C1_3D",
    y="pca_C2_3D",
    z="pca_C3_3D",  # O novo eixo Z
    # Define a coluna que será usada para colorir os pontos (os clusters)
    color='Cliente_Novo',
    # Você pode usar a paleta "hls" do Plotly, ou outras
    color_discrete_sequence=px.colors.qualitative.Alphabet,
    # Texto que aparece ao passar o mouse sobre os pontos
    hover_data=['Cliente_Novo'],
    title=f'Visualização 3D Interativa de Clusters com PCA do Cluster {cluster_resultado}'
)

# --- Ajuste do Tamanho dos Pontos ---
# O parâmetro 'size' dentro de 'marker' controla o diâmetro dos pontos em pixels.
# O valor padrão costuma ser 5 ou 6. Diminuir para 3 ou 4 torna os pontos menores e mais nítidos.
fig.update_traces(
    marker=dict(
        size=4,  # <-- Reduz o tamanho dos pontos
        opacity=0.8 # Mantém a opacidade (opcional)
    ),
    selector=dict(mode='markers') # Garante que apenas os marcadores sejam afetados
)

# --- Personalização do Layout 3D ---
fig.update_layout(
    scene=dict(
        xaxis_title='Componente PCA 1',
        yaxis_title='Componente PCA 2',
        zaxis_title='Componente PCA 3'
    ),
    height=700,
    width=900
)

# Mostra o gráfico interativo
fig.show()
```

```python
# Definindo função de agregação pela moda
def agg_moda(x):
    m = x.dropna().mode()
    if not m.empty:
        return m.iloc[0]
    return np.nan
```

```python
# Definindo função de agregação pelo top 5 da moda
def top5_comuns(x):
    return x.dropna().value_counts().head(5).index.tolist()
```

```python
# Agregando os dados por cluster 5, utilizando a média para as colunas numéricas e a moda para as colunas categóricas
df_agrupado = df.groupby('CLUSTER_5').agg(
    QTD_CLIENTES=('CD_CLIENTE', 'nunique'),
    DS_PROD_moda=('DS_PROD_MODA', agg_moda),
    DS_PROD_top5=('DS_PROD_MODA', top5_comuns),
    DS_LIN_REC_moda=('DS_LIN_REC_MODA', agg_moda),
    DS_LIN_REC_top5=('DS_LIN_REC_MODA', top5_comuns),
    CIDADE_moda=('CIDADE', agg_moda),
    CIDADE_top5=('CIDADE', top5_comuns),
    DS_CNAE_moda=('DS_CNAE', agg_moda),
    DS_CNAE_top5=('DS_CNAE', top5_comuns),
    DS_SEGMENTO_moda=('DS_SEGMENTO', agg_moda),
    DS_SEGMENTO_top5=('DS_SEGMENTO', top5_comuns),
    DS_SUBSEGMENTO_moda=('DS_SUBSEGMENTO', agg_moda),
    DS_SUBSEGMENTO_top5=('DS_SUBSEGMENTO', top5_comuns),
    MARCA_TOTVS_moda=('MARCA_TOTVS_MODA', agg_moda),
    MARCA_TOTVS_top5=('MARCA_TOTVS_MODA', top5_comuns),
    MODAL_COMERC_moda=('MODAL_COMERC_MODA', agg_moda),
    MODAL_COMERC_top5=('MODAL_COMERC_MODA', top5_comuns),
    PAIS_moda=('PAIS', agg_moda),
    PAIS_top5=('PAIS', top5_comuns),
    PERIODICIDADE_moda=('PERIODICIDADE_MODA', agg_moda),
    PERIODICIDADE_top5=('PERIODICIDADE_MODA', top5_comuns),
    UF_moda=('UF', agg_moda),
    UF_top5=('UF', top5_comuns),
    VL_TOTAL_CONTRATO_sum=('VL_TOTAL_CONTRATO_SOMA', 'sum'),
    VL_TOTAL_CONTRATO_median=('VL_TOTAL_CONTRATO_SOMA', 'median'),
    VL_TOTAL_CONTRATO_mean=('VL_TOTAL_CONTRATO_SOMA', 'mean'),
    VL_TOTAL_CONTRATO_min=('VL_TOTAL_CONTRATO_SOMA', 'min'),
    VL_TOTAL_CONTRATO_max=('VL_TOTAL_CONTRATO_SOMA', 'max'),
    FAIXA_VLR_TOT_SOMA_CONTRATOS_moda=('FAIXA_VLR_TOT_SOMA_CONTRATOS', agg_moda),
    FAIXA_VLR_TOT_SOMA_CONTRATOS_top5=('FAIXA_VLR_TOT_SOMA_CONTRATOS', top5_comuns),
    DIAS_CLIENTE_median=('DIAS_CLIENTE', 'median'),
    DIAS_CLIENTE_mean=('DIAS_CLIENTE', 'mean'),
    DIAS_CLIENTE_min=('DIAS_CLIENTE', 'min'),
    DIAS_CLIENTE_max=('DIAS_CLIENTE', 'max'),
    QTD_CONTRATACOES_12M_sum=('QTD_CONTRATACOES_12M', 'sum'),
    QTD_CONTRATACOES_12M_median=('QTD_CONTRATACOES_12M', 'median'),
    QTD_CONTRATACOES_12M_mean=('QTD_CONTRATACOES_12M', 'mean'),
    QTD_CONTRATACOES_12M_min=('QTD_CONTRATACOES_12M', 'min'),
    QTD_CONTRATACOES_12M_max=('QTD_CONTRATACOES_12M', 'max'),
    VLR_CONTRATACOES_12M_sum=('VLR_CONTRATACOES_12M', 'sum'),
    VLR_CONTRATACOES_12M_median=('VLR_CONTRATACOES_12M', 'median'),
    VLR_CONTRATACOES_12M_mean=('VLR_CONTRATACOES_12M', 'mean'),
    VLR_CONTRATACOES_12M_min=('VLR_CONTRATACOES_12M', 'min'),
    VLR_CONTRATACOES_12M_max=('VLR_CONTRATACOES_12M', 'max'),
).reset_index()
```
